## 1. Conversion formulas

We will use the following mathematical formulas:

**Celsius to Fahrenheit:**

$$F = (C \times \frac{9}{5}) + 32$$

**Fahrenheit to Celsius:**

$$C = (F - 32) \times \frac{5}{9}$$

## 2. Define conversion functions

We will create two reusable functions, one for each conversion direction.

```python
def celsius_to_fahrenheit(celsius: float) -> float:
    """Convert a temperature from Celsius to Fahrenheit."""
    return (celsius * 9 / 5) + 32


def fahrenheit_to_celsius(fahrenheit: float) -> float:
    """Convert a temperature from Fahrenheit to Celsius."""
    return (fahrenheit - 32) * 5 / 9
```

## 3. Test the conversion functions

Let's test our functions using several common Celsius temperatures.

```python
temperatures_celsius = [0, 20, 25, 37, 100]

for temperature in temperatures_celsius:
    converted = celsius_to_fahrenheit(temperature)
    print(f"{temperature:>5.1f} °C = {converted:>5.1f} °F")
```

## 4. Convert an individual temperature

Change the value of `temperature_c` below to try another conversion.

```python
temperature_c = 25
temperature_f = celsius_to_fahrenheit(temperature_c)

print(f"{temperature_c} °C is equal to {temperature_f:.1f} °F")
```

## 5. Validate the results

Finally, we will use assertions to check that the conversion functions return the expected results.

```python
assert celsius_to_fahrenheit(0) == 32
assert celsius_to_fahrenheit(100) == 212
assert fahrenheit_to_celsius(32) == 0
assert fahrenheit_to_celsius(212) == 100

print("All conversion checks passed successfully.")
```
