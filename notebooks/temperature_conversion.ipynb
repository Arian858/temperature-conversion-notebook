## Conversion Reference

**Celsius to Fahrenheit**

F = (C × 9/5) + 32

**Fahrenheit to Celsius**

C = (F − 32) × 5/9

```python

from math import isfinite


def validate_temperature(value, parameter_name):
    """Validate that a temperature is a finite numeric value."""

    if isinstance(value, bool) or not isinstance(value, (int, float)):
        raise TypeError(
            f"{parameter_name} must be an integer or float."
        )

    if not isfinite(value):
        raise ValueError(
            f"{parameter_name} must be a finite number."
        )

    return float(value)


def celsius_to_fahrenheit(celsius: float) -> float:
    """Convert a temperature from Celsius to Fahrenheit."""

    celsius = validate_temperature(celsius, "Celsius")
    return (celsius * 9 / 5) + 32


def fahrenheit_to_celsius(fahrenheit: float) -> float:
    """Convert a temperature from Fahrenheit to Celsius."""

    fahrenheit = validate_temperature(fahrenheit, "Fahrenheit")
    return (fahrenheit - 32) * 5 / 9
```


## 3. Conversion Examples

This section evaluates the conversion functions using a selection of commonly encountered temperatures.

The examples demonstrate both conversion directions and provide a simple way to verify the calculated results.


```python

# Example temperatures in Celsius
celsius_examples = [0, 20, 25, 37, 100]

print("CELSIUS TO FAHRENHEIT")
print("-" * 30)
print(f"{'Celsius (°C)':>14} {'Fahrenheit (°F)':>16}")
print("-" * 30)

for temperature in celsius_examples:
    result = celsius_to_fahrenheit(temperature)
    print(f"{temperature:>14.1f} {result:>16.1f}")


# Example temperatures in Fahrenheit
fahrenheit_examples = [32, 68, 77, 98.6, 212]

print("\nFAHRENHEIT TO CELSIUS")
print("-" * 30)
print(f"{'Fahrenheit (°F)':>16} {'Celsius (°C)':>14}")
print("-" * 30)

for temperature in fahrenheit_examples:
    result = fahrenheit_to_celsius(temperature)
    print(f"{temperature:>16.1f} {result:>14.1f}")
```


## 4. Input Validation and Automated Testing

Reliable conversion functions should produce accurate results and reject invalid inputs.

This section uses automated assertions to verify the expected behavior of the conversion functions and their input validation mechanisms.


```python

import math

# Test 1: Celsius to Fahrenheit
assert math.isclose(celsius_to_fahrenheit(0), 32.0)
assert math.isclose(celsius_to_fahrenheit(100), 212.0)

# Test 2: Fahrenheit to Celsius
assert math.isclose(fahrenheit_to_celsius(32), 0.0)
assert math.isclose(fahrenheit_to_celsius(212), 100.0)

# Test 3: Decimal temperature conversion
assert math.isclose(fahrenheit_to_celsius(98.6), 37.0)

# Test 4: Invalid input type
try:
    celsius_to_fahrenheit("hello")
except TypeError:
    pass
else:
    raise AssertionError("Invalid input type was not rejected.")

# Test 5: Non-finite numeric input
try:
    celsius_to_fahrenheit(float("inf"))
except ValueError:
    pass
else:
    raise AssertionError("Non-finite input was not rejected.")

print("All automated tests passed successfully.")
```


## 5. Interactive Temperature Converter

This section provides an interactive interface that allows users to select a conversion direction and enter a temperature value.

The program calculates and displays the converted temperature while handling invalid inputs.


```python

print("=" * 45)
print("       TEMPERATURE CONVERSION TOOL")
print("=" * 45)

print("\nAvailable conversions:")
print("C - Celsius to Fahrenheit")
print("F - Fahrenheit to Celsius")

direction = input("\nSelect conversion direction (C/F): ").strip().upper()

if direction == "C":

    raw_temperature = input("Enter temperature in Celsius: ")

    try:
        temperature = float(raw_temperature)
        result = celsius_to_fahrenheit(temperature)

    except (ValueError, TypeError):
        print("\nError: Please enter a valid finite number.")

    else:
        print("\nCONVERSION RESULT")
        print("-" * 30)
        print(f"Celsius:    {temperature:.2f} °C")
        print(f"Fahrenheit: {result:.2f} °F")

elif direction == "F":

    raw_temperature = input("Enter temperature in Fahrenheit: ")

    try:
        temperature = float(raw_temperature)
        result = fahrenheit_to_celsius(temperature)

    except (ValueError, TypeError):
        print("\nError: Please enter a valid finite number.")

    else:
        print("\nCONVERSION RESULT")
        print("-" * 30)
        print(f"Fahrenheit: {temperature:.2f} °F")
        print(f"Celsius:    {result:.2f} °C")

else:
    print("\nError: Invalid selection. Please enter C or F.")
```
